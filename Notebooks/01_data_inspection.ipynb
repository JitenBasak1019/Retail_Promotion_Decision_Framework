```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path("D:/DataProject/Data/Raw")

csv_files = list(DATA_PATH.glob("*.csv"))

print("CSV files found:")

for file in csv_files:
    print("-", file.name)

print("\nNumber of CSV files:", len(csv_files))
```

```python
for file in csv_files:
    df = pd.read_csv(file)

    print("\n" + "=" * 60)
    print(f"FILE: {file.name}")
    print("=" * 60)

    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])

    print("\nColumn names:")
    print(df.columns.tolist())
```

```python
import pandas as pd

files_to_check = [
    "campaign_desc.csv",
    "campaign_table.csv",
    "coupon.csv",
    "coupon_redempt.csv",
    "hh_demographic.csv",
    "product.csv",
    "transaction_data.csv"
]

dataframes = {}

for filename in files_to_check:
    print({filename})
    
    df = pd.read_csv(DATA_PATH / filename)
    dataframes[filename] = df
    
    print(f"Shape: {df.shape}")
    print()
```

```python
for filename, df in dataframes.items():
    
    print("\n" + "_" * 70)
    print(filename)
    
    missing = df.isnull().sum()
    
    print(missing[missing > 0])
    print("-" * 70)
```

```python
for filename, df in dataframes.items():
    
    print("\n" + "-" * 70)
    print(filename)
    print(df.dtypes)
    print("-" * 70)
```

```python
for filename, df in dataframes.items():
    
    print(f"{filename}: {df.duplicated().sum()} duplicate rows")
```

```python
#Investigating the duplicate values in coupon table
coupon_df = dataframes["coupon.csv"]

duplicate_coupon_rows = coupon_df[
    coupon_df.duplicated(keep=False)
].sort_values(
    by=["COUPON_UPC", "PRODUCT_ID", "CAMPAIGN"]
)

print("Number of duplicate rows:", len(duplicate_coupon_rows))

duplicate_coupon_rows.head(20)
```

```python
print(
    "Unique COUPON_UPC values:",
    coupon_df["COUPON_UPC"].nunique()
)

print(
    "Unique PRODUCT_ID values:",
    coupon_df["PRODUCT_ID"].nunique()
)

print(
    "Unique CAMPAIGN values:",
    coupon_df["CAMPAIGN"].nunique()
)
```

```python
print(
    coupon_df.groupby(
        ["COUPON_UPC", "PRODUCT_ID", "CAMPAIGN"]
    ).size().sort_values(ascending=False).head(20)
)
```

```python
coupon_df = dataframes["coupon.csv"]

print("Original rows:", len(coupon_df))
print("Duplicate rows:", coupon_df.duplicated().sum())
print("Rows after removing exact duplicates:",
      len(coupon_df.drop_duplicates()))

print(
    "Rows removed:",
    len(coupon_df) - len(coupon_df.drop_duplicates())
)
```

```python
# Removing the duplicates from coupon.csv and making a new file with cleaned rows
coupon_df = dataframes["coupon.csv"]

coupon_clean = coupon_df.drop_duplicates().copy()

print("cleaned_table_rows :", len(coupon_clean))
```

```python
processed_path = Path(r"D:\DataProject\Data\Processed")

processed_path.mkdir(parents=True, exist_ok=True)

coupon_clean.to_csv(
    processed_path / "coupon_clean.csv",
    index=False
)

print("Saved:", processed_path / "coupon_clean.csv")
```
